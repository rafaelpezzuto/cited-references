```python
with open('issn_to_all.v0.8b.countries_formatted.compat.csv', 'w') as fout:
    fout.write('|'.join(b6[0].keys()) + '\n')
    for i in b8:
        els = [
            i['GOLD_ISSN'],# 'ISSNL' 
            i['PORTAL_ISSN_2019_MAIN_TITLE'] or i['PORTAL_ISSN_2019_MAIN_ABBREVIATED_TITLE'] or i['PORTAL_ISSN_2024_MAIN_TITLE'] or i['TITLES'].split('#')[0],# 'MAIN_TITLE' 
            i['PORTAL_ISSN_2019_MAIN_ABBREVIATED_TITLE'],# 'MAIN_ABBREV_TITLE' 
            i['ISSNS'],# 'ISSNs' 
            i['TITLES'],# 'OTHER_TITLEs' 
            str(int(i['IS_PORTAL_ISSN_2019']) or int(i['IS_PORTAL_ISSN_2024'])),# 'PORTAL_ISSN' 
            i['IS_DOAJ'],# 'DOAJ' 
            i['IS_LATINDEX'],# 'LATINDEX' 
            i['IS_NLM'],# 'NLM' 
            i['IS_SCIELO'],# 'SCIELO' 
            i['IS_SCIMAGOJR'],# 'SCIMAGO_JR' 
            str(int(i['IS_SCOPUS_ACCEPTED']) or int(i['IS_SCOPUS_SOURCES'])),# 'SCOPUS' 
            i['IS_ULRICH'],# 'ULRICH'
            i['IS_WOS_EXTRA'],# 'WOS'
            i['IS_WOS_JCR'],# 'WOS_JCR'
            i['COUNTRY'],# 'COUNTRIES'
            '',# 'YEARS'
        ]
        fout.write('|'.join(els) + '\n')
```
